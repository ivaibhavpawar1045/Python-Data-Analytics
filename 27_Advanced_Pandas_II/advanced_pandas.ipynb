```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" , encoding="latin1")
df.shape
```

```python
df.info()
```

### apply()

```python
def classify_profit(profit):
     if profit > 100:
          return "Highly Profitable"
     elif profit > 0:
          return "Profitable"
     else:
          return "Loss Making"

df["Profit_Category"] = df["Order Profit Per Order"].apply(
     classify_profit
)
df[[
    "Order Profit Per Order",
    "Profit_Category"
]].head(10)
```

```python
df["Profit_Category"].value_counts()
```

### map()

```python
df["Delivery_Risk_Status"] = df["Late_delivery_risk"].map(
     {
          0:"No Risk" ,
          1:"Late Risk"
     }
)
df[[
     "Late_delivery_risk",
     "Delivery_Risk_Status"
]].head(10)
```

```python
df["Delivery_Risk_Status"].value_counts()
```
