```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" , encoding="latin1")
df.shape
```

```python
df[
     ["Sales" ,
      "Order Profit Per Order" ,
      "Customer Segment" ,
      "Market" ,
      "Late_delivery_risk",
      "Shipping Mode"]
].head()
```

```python
df.info()
```

### apply()

```python
def classify_profit(profit):
     if profit > 100:
          return "Highly Profitable"
     elif profit > 0:
          return "Profitable"
     else:
          return "Loss Making"

df["Profit_Category"] = df["Order Profit Per Order"].apply(
     classify_profit
)
df[[
    "Order Profit Per Order",
    "Profit_Category"
]].head(10)
```

```python
df["Profit_Category"].value_counts()
```

### map()

```python
df["Delivery_Risk_Status"] = df["Late_delivery_risk"].map(
     {
          0:"No Risk" ,
          1:"Late Risk"
     }
)
df[[
     "Late_delivery_risk",
     "Delivery_Risk_Status"
]].head(10)
```

```python
df["Delivery_Risk_Status"].value_counts()
```

### replace(), where(), query()

### 1) replace() 

```python
df["Customer Segment"].value_counts()
```

```python
df["Customer_Segment_Standard"] = df["Customer Segment"].replace({
    "Consumer": "B2C",
    "Corporate": "B2B",
    "Home Office": "Home Office"
})
df["Customer_Segment_Standard"].value_counts()
```

### 2) where()

```python
df["Profit_Positive"] = df["Order Profit Per Order"].where(
    df["Order Profit Per Order"] > 0,
    0
)
df[[
    "Order Profit Per Order",
    "Profit_Positive"
]].head(10)
```

### 3) query()

```python
high_value_orders = df.query(
    "Sales > 1000 and `Order Profit Per Order` > 100"
)

high_value_orders.head()
```
