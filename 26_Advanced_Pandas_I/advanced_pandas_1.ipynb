```python
import pandas as pd
```

```python
# Orders
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Customer_ID": ["C01", "C02", "C01", "C04"],
    "Sales": [500, 700, 300, 900]
})

orders
```

```python
# Customers
customers = pd.DataFrame({
    "Customer_ID": ["C01", "C02", "C03"],
    "Customer_Name": ["Rahul", "Amit", "Sneha"],
    "Segment": ["Consumer", "Corporate", "Consumer"]
})

customers
```

```python
# merge()
merged_data = pd.merge(
     orders , customers , on="Customer_ID"
)
merged_data
```

```python
# Inner Join
pd.merge(
     orders , customers , on="Customer_ID" , how="inner"
).shape
```

```python
# Outer Join
pd.merge(
     orders , customers , on="Customer_ID" , how="outer"
).shape
```

```python
# Left join
pd.merge(
     orders , customers , on="Customer_ID" , how="left"
).shape
```

```python
# Right Join
pd.merge(
     orders , customers , on="Customer_ID" , how="right"
).shape
```

### Vertical concat() 

```python
# Create two monthly datasets
# January Sales
sales_jan = pd.DataFrame({
    "Order_ID": [101, 102],
    "Sales": [500, 700]
})
```

```python
# February Sales
sales_feb = pd.DataFrame({
    "Order_ID": [103, 104],
    "Sales": [300, 900]
})
```

```python
sales = pd.concat(
     [sales_jan , sales_feb],
     ignore_index=True
)
sales
```

```python
sales.shape
```

### Horizontal concat()

```python
# create another DataFrame
customer_info = pd.DataFrame({
    "Customer_ID": ["C01", "C02", "C01", "C04"],
    "Segment": ["Consumer", "Corporate", "Consumer", "Home Office"]
})
```

```python
sales_with_customer = pd.concat(
    [sales, customer_info],
    axis=1
)

sales_with_customer
```

```python
sales_with_customer.shape
```

### join()

```python
orders_indexed = orders.set_index("Customer_ID")
customers_indexed = customers.set_index("Customer_ID")

orders_indexed
```

```python
customers_indexed
```

```python
# Perform left join
joined_data = orders_indexed.join(
     customers_indexed ,
     how= "left"
)
joined_data
```

```python
joined_data.shape
```

### BUSINESS SCENARIOS

```python
customer_sales = merged_data.groupby(
     "Segment"
).agg(
     Total_sales = ("Sales" , "sum"),
     Total_orders = ("Order_ID" , "nunique")
)
customer_sales
```

```python
     # calculate each segment
customer_sales["Sales_Contribution_%"] = (
    customer_sales["Total_sales"]
     .div(customer_sales["Total_sales"].sum())
     .mul(100)
)
customer_sales
```
