# DAY 32 : ADVANCED MATPLOTLIB

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
months = np.array(["Jan" , "Feb" , "Mar" , "Apr" , "May" , "Jun"])
sales = np.array([12000 , 150000 , 13500 , 18000 , 21000 , 19500])
profit = np.array([1800 , 2400 , 2000 , 3100 , 3800 , 3400])
```

### Subplots and Figure Layout

```python
# Create two charts in one figure
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(months, sales, marker="o")
axes[0].set_title("Monthly Sales")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Sales (₹)")
axes[0].grid(True, alpha=0.3)

axes[1].bar(months, profit)
axes[1].set_title("Monthly Profit")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Profit (₹)")

plt.tight_layout()
plt.show()
```

```python
# Create a 2 × 2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(months, sales, marker="o")
axes[0, 0].set_title("Sales Trend")

axes[0, 1].bar(months, profit)
axes[0, 1].set_title("Monthly Profit")

axes[1, 0].scatter(sales, profit)
axes[1, 0].set_title("Sales vs Profit")
axes[1, 0].set_xlabel("Sales (₹)")
axes[1, 0].set_ylabel("Profit (₹)")

axes[1, 1].bar(months, sales - profit)
axes[1, 1].set_title("Sales After Profit")
axes[1, 1].set_ylabel("Sales − Profit (₹)")

plt.tight_layout()
plt.show()
```

```python
# Find the month with the highest profit:
best_index = np.argmax(profit)

print("Highest Profit Month:", months[best_index])
print("Highest Profit:", profit[best_index])
```

### Advanced Customization and Annotations

```python
# . Add an annotation to the highest-sales month
best_sales_index = np.argmax(sales)

plt.figure(figsize=(8, 4))
plt.plot(months, sales, marker="o")

plt.annotate(
    "Highest Sales",
    xy=(months[best_sales_index], sales[best_sales_index]),
    xytext=(10, -30),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Sales with Peak Highlighted")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Format the y-axis with commas
import matplotlib.ticker as ticker

plt.figure(figsize=(8, 4))
plt.bar(months, sales)

plt.gca().yaxis.set_major_formatter(
    ticker.StrMethodFormatter("{x:,.0f}")
)

plt.title("Monthly Sales — Formatted Axis")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.show()
```

```python
# Highlight months with profit of at least ₹3,000
high_profit = profit >= 3000

plt.figure(figsize=(8, 4))
plt.bar(months, profit)

plt.axhline(
    y=3000,
    linestyle="--",
    label="Profit Target: ₹3,000"
)

plt.title("Monthly Profit vs Target")
plt.xlabel("Month")
plt.ylabel("Profit (₹)")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()

print("Months meeting target:", months[high_profit])
```

### Stacked Bar, Pie Chart and Box Plot

```python
# Create product-category data
categories = np.array(["Electronics", "Clothing", "Furniture", "Food"])

online_sales = np.array([12000, 9000, 15000, 7000])
store_sales = np.array([10000, 11000, 13000, 8000])
```

```python
# Stacked bar chart
plt.figure(figsize=(8, 4))

plt.bar(categories, online_sales, label="Online")
plt.bar(categories, store_sales, bottom=online_sales, label="Store")

plt.title("Sales by Category and Channel")
plt.xlabel("Category")
plt.ylabel("Sales (₹)")
plt.legend()
plt.show()
```

```python
# Pie chart — sales contribution
category_totals = online_sales + store_sales

plt.figure(figsize=(7, 7))

plt.pie(
    category_totals,
    labels=categories,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Category Contribution to Total Sales")
plt.show()
```

```python
# Box plot — order-value distribution
order_values = np.array([
    200, 250, 300, 350, 400,
    450, 500, 550, 600, 650,
    700, 750, 800, 850, 2000
])

plt.figure(figsize=(7, 4))

plt.boxplot(order_values)

plt.title("Order Value Distribution")
plt.ylabel("Order Value (₹)")
plt.show()
```

```python
# Identify the highest-revenue category and its share of total sales.
best_category_index = np.argmax(category_totals)

print("Top Category:", categories[best_category_index])
print("Category Sales:", category_totals[best_category_index])
print(
    "Contribution (%):",
    round(category_totals[best_category_index] / category_totals.sum() * 100, 2)
)
```
