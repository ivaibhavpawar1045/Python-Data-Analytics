```python
# remove the duplicate
clean_df = df.drop_duplicates().copy()
clean_df.shape
```

```python
# checks the datatypes
clean_df.dtypes
```

```python
clean_df["Order Date"].head()
```

```python
clean_df["Order Date"].dtype
```

```python
# convert order date to date datatype
clean_df["Order Date"] = pd.to_datetime(
    clean_df["Order Date"],
    format="mixed"
)
```

```python
clean_df["Order Date"].dtype
```

```python
clean_df["Order Date"].head()
```

```python
# Extract the date information
clean_df["Year"] = clean_df["Order Date"].dt.year
clean_df["Month"] = clean_df["Order Date"].dt.month
clean_df["Month Name"] = clean_df["Order Date"].dt.month_name()
clean_df["Day Name"] = clean_df["Order Date"].dt.day_name()
```

```python
clean_df.head()
```

BUSINESS ANALYSIS

```python
# yearly highest sales
yearly_sales = clean_df.groupby("Year")["Sales"].sum()
yearly_sales
```

```python
yearly_sales.idxmax(),yearly_sales.max()
```

```python
# monthly highest sales
monthly_sales = clean_df.groupby(
     "Month Name"

)["Sales"].sum()
monthly_sales
```

```python
# Highest profit generated year
yearly_profit = clean_df.groupby("Year")["Sales"].sum()
yearly_profit
```

```python
yearly_profit.idxmax(),yearly_profit.max()
```

```python
# which day has more orders
day_orders = clean_df.groupby("Day Name")["Sales"].sum()
day_orders
```

```python
# which category generated the highest sales
category_sales = clean_df.groupby("Category")["Sales"].sum()
category_sales
```

```python
category_sales.idxmax(),category_sales.max()
```

```python
# region has highest sales
region_sales = clean_df.groupby(
    "Region"
)["Sales"].sum()

region_sales
```

```python
region_sales.idxmax(), region_sales.max()
```

MONTHLY SALES TRENDS
