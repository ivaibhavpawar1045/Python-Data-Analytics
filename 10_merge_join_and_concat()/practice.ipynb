MERGE() , JOIN() , CONCAT()

```python
import pandas as pd
```

```python
employee_df = pd.DataFrame({
     "Employee_ID" : [101,102,103,104],
     "Employee" : ["Rahul" , "Priya" , "Amit" , "Sneha"],
     "Department_ID" : [1,2,3,2]
})

employee_df
```

```python
department_df = pd.DataFrame({
    "Department_ID": [1, 2, 3],
    "Department": ["Sales", "HR", "IT"]
})

department_df
```

```python
merge_df = pd.merge(
     employee_df,
     department_df,
     on="Department_ID"
)

merge_df
```

TYPES OF MERGE ;- INNER JOIN , LEFT JOIN , RIGHT JOIN , OUTER JOIN

```python
employee_df = pd.DataFrame({
    "Employee_ID":[101,102,103,104],
    "Employee":["Rahul","Priya","Amit","Sneha"],
    "Department_ID":[1,2,3,5]
})

employee_df
```

```python
department_df = pd.DataFrame({
    "Department_ID":[1,2,3,4],
    "Department":["Sales","HR","IT","Finance"]
})

department_df
```

```python
inner_df = pd.merge(
     employee_df , department_df ,
     on="Department_ID" , 
     how="inner"
)
inner_df
```

```python
left_df = pd.merge(
     employee_df , department_df,
     on="Department_ID",
     how="left"
)
left_df
```

```python
right_df = pd.merge(
     employee_df , department_df,
     on="Department_ID",
     how="right"
)
right_df
```

```python
outer_df = pd.merge(
     employee_df , department_df,
     on="Department_ID",
     how="outer"
)
outer_df
```

CONCAT ()    -
          BY ROW WISE
          BY COLUMN WISE

```python
employee1 = pd.DataFrame({
    "Employee":["Rahul","Priya"],
    "Salary":[50000,65000]
})

employee1
```

```python
employee2 = pd.DataFrame({
    "Employee":["Amit","Sneha"],
    "Salary":[70000,55000]
})

employee2
```

```python
# conact() by row wise
pd.concat(
     [employee1 , employee2],
     ignore_index=False
)
```

```python
# concat() by column wise
employee_info = pd.DataFrame({
    "Department":["Sales","HR"]
})

pd.concat(
    [employee1, employee_info],
    axis=1 
)
```

JOIN() 

```python
employee_df = pd.DataFrame(
    {
        "Employee":["Rahul","Priya"],
        "Salary":[50000,65000]
    },
    index=[1,2]
)
employee_df
```

```python
department_df = pd.DataFrame({
     "Department":["Sales" , "HR"]
} , index=[1,2]
)
department_df
```

```python
employee_df.join(department_df)
```
