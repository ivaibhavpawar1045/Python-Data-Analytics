```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import matplotlib

matplotlib.__version__
```

```python
# create business data
days = np.array(["Mon" , "Tue" , "Wed" , "Thu" , "Fri" , "Sat" , "Sun"])

sales = np.array([1200 , 1500 , 1100 , 1800 , 2100 , 1900 , 2500])
```

```python
# Create first line chart
plt.figure(figsize=(8,4))
plt.plot(days, sales)
plt.title("Daily Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show
```

```python
# Identify the highest-sales day
highest_index = np.argmax(sales)

print("Highest Sales Day:", days[highest_index])
print("Highest Sales:", sales[highest_index])
```

### Bar Charts


```python
# compare sales across seven days?
plt.figure(figsize=(8, 4))

plt.bar(days, sales)

plt.title("Daily Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
# horizontal bar chart
plt.figure(figsize=(8, 4))

plt.barh(days, sales)

plt.title("Daily Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")

plt.show()
```

```python
# customize bar graph
plt.figure(figsize=(8,4))
plt.bar(
     days ,
     sales ,
     label = "Daily Sales"
)
plt.title("Weekly Sales Performance")
plt.xlabel("Day")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(axis="y" , linestyle="--" , alpha= 0.5)
plt.show
```

```python
# Calculate the average daily sales and identify the days above average.
avg_sales = np.mean(sales)

print("Average Daily Sales" , avg_sales)
print("Days Above Average" , days[sales > avg_sales])
```

### Scatter Plot, Histogram & Customization

### 1. Scatter Plot

```python
# Do higher customer counts correspond to higher sales?
customers = np.array([20, 25, 18, 30, 35, 32, 40])

plt.figure(figsize=(8, 4))

plt.scatter(customers, sales)

plt.title("Customer Count vs Sales")
plt.xlabel("Number of Customers")
plt.ylabel("Sales (₹)")

plt.show()
```

### 2. Histogram

```python
# How are daily sales distributed across the week?
plt.figure(figsize=(8, 4))

plt.hist(sales, bins=5, edgecolor="black")

plt.title("Distribution of Daily Sales")
plt.xlabel("Sales (₹)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.scatter(
    customers,
    sales,
    marker="o",
    label="Daily Performance"
)

plt.title("Customer Count vs Daily Sales")
plt.xlabel("Number of Customers")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
# Find the sales per customer for each day.
sales_per_customer = sales / customers

plt.figure(figsize=(8, 4))

plt.bar(days, sales_per_customer)

plt.title("Sales per Customer by Day")
plt.xlabel("Day")
plt.ylabel("Sales per Customer (₹)")

plt.show()
```

```python
# Identify the highest sales-per-customer day:
best_index = np.argmax(sales_per_customer)

print("Best Day:", days[best_index])
print("Sales per Customer:", round(sales_per_customer[best_index], 2))
```
