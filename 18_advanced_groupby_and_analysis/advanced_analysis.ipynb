ADVANCED GROUPBY AND BUSINESS ANALYSIS

```python
import pandas as pd
```

```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" , encoding="latin1")
df.head()
```

MULTIPLE AGGREGATIONS

```python
# find the category product with there total sales and profit as well as average sales and profit
product_category = (
     df.groupby("Category Name")
     [["Sales" , "Order Profit Per Order"]]
     .agg(["sum" , "mean"])
)
product_category.head()
```

```python
high_profit = (
     df.groupby("Category Name")["Order Profit Per Order"]
     .sum().sort_values(ascending=False)
)
high_profit.head(1)
```

```python
high_sales = (
     df.groupby("Category Name")["Sales"]
     .sum().sort_values(ascending=False)
)
high_sales.head(1)
```

```python
high_avg_profit = (
     df.groupby("Category Name")["Order Profit Per Order"]
     .mean().sort_values(ascending=False)
)
high_avg_profit.head(1)
```

NAMED AGGREGATIONS

```python
category_summary = df.groupby("Category Name").agg(
     Total_Sales = ("Sales" , "sum"),
     Average_Sales = ("Sales" , "mean"),
     Total_Profit = ("Order Profit Per Order" , "sum"),
     Average_Profit = ("Order Profit Per Order" , "mean"),
     No_of_Order = ("Order Id" , "count")
)
category_summary.head()
```

```python
category_summary.sort_values(
     by="Total_Profit",
     ascending=False
).head(1)
```

```python
category_summary.sort_values(
     by="Average_Profit",
     ascending=False
).head(1)
```

GROUPED BUSINESS METRICS

```python
market_segment_metrics = (
     df.groupby(["Market" , "Customer Segment"])["Sales"]
     .sum().reset_index()
)
market_segment_metrics
```

```python
# which customer segment has the highest sales within each market
market_segment_metrics.loc[
     market_segment_metrics.groupby("Market")["Sales"].idxmax()
]
```

RANKING WITHIN GROUPS

```python
# for each market , rank the customer segment acco. to their sales 
market_segment_metrics["Sales Rank"] =(
     market_segment_metrics.groupby("Market")["Sales"]
     .rank(
          ascending=False ,
          method="dense"
     )
)
```

```python
market_segment_metrics.sort_values(
     ["Market" , "Sales Rank"]
)
```

BUSINESS CHALLENGE :-
                    WHICH CUSTOMER SEGMENT HAS HIGHEST AVERAGE ORDER VLAUE IN EACH MARKET

```python
# find the average market segment
average_market_segmet = (
     df.groupby(["Market" , "Customer Segment"])["Sales"]
     .mean().reset_index(name="Average Sales")
)
average_market_segmet
```

```python
# find the highest sales in every market
average_market_segmet.loc[
     average_market_segmet.groupby("Market")["Average Sales"].idxmax()
]
```
