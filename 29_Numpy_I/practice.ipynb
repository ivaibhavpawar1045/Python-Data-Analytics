### BUSINESS CHALLENGES

Sales ≥ 1,800 → "High"



Sales ≥ 1,400 → "Medium"



Otherwise → "Low"

```python
sales_category = np.select(
     [
          daily_sales >= 1800 ,
          daily_sales >= 1400
     ],
     [
          "High",
          "Medium"
     ],
     default="Low"
)
sales_category
```

count each category

```python
np.unique(sales_category, return_counts=True)
```

```python
# Calculate the percentage of total sales by category

for category in np.unique(sales_category):
     category_total = daily_sales[sales_category == category].sum()
     contribution = category_total / daily_sales.sum() * 100

     print(category , category_total , round(contribution , 2))
```
