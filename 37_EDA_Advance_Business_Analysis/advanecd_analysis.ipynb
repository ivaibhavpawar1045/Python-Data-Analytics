```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Setup the dataset
sales_data_clean = pd.read_excel("online_retail_II_cleaned.xlsx")
sales_data_clean.shape
```

```python
# Step 1: Summarize transaction sales
sales_data_clean["Sales"].describe()
```

```python
# Step 2: Compare typical sales with extreme values
print("Mean sales:", sales_data_clean["Sales"].mean())
print("Median sales:", sales_data_clean["Sales"].median())
print("Minimum sales:", sales_data_clean["Sales"].min())
print("Maximum sales:", sales_data_clean["Sales"].max())
```

```python
# Step 3: Examine the distribution's percentiles
sales_data_clean["Sales"].quantile(
    [0.01, 0.25, 0.50, 0.75, 0.95, 0.99]
)
```

```python
# Step 4A — Create the full-range histogram
# Visualize the complete sales distribution
plt.figure(figsize=(8, 4))

sns.histplot(
    data=sales_data_clean,
    x="Sales",
    bins=50
)

plt.title("Distribution of Transaction Sales")
plt.xlabel("Transaction Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Step 4B — Zoom into the central range
# Focus on the central transaction range
plt.figure(figsize=(8, 4))

sns.histplot(
    data=sales_data_clean,
    x="Sales",
    bins=50
)

plt.xlim(-100, 500)
plt.title("Transaction Sales: Central Range")
plt.xlabel("Transaction Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Step 4C — Inspect potential outliers
plt.figure(figsize=(8, 4))

sns.boxplot(data=sales_data_clean, x="Sales")

plt.title("Box Plot of Transaction Sales")
plt.xlabel("Transaction Sales")
plt.show()
```

```python
# Focus on sales between -100 and 500 for visualization
sales_central = sales_data_clean.loc[
    sales_data_clean["Sales"].between(-100, 500),
    "Sales"
]

plt.figure(figsize=(8, 4))
sns.histplot(x=sales_central, bins=50)

plt.title("Transaction Sales: Central Range")
plt.xlabel("Transaction Sales")
plt.ylabel("Frequency")
plt.show()
```

### Bivariate analysis : It examines the relationship between two variables

```python
# Relationship between quantity and sales
plt.figure(figsize=(8, 4))

sns.scatterplot(
    data=sales_data_clean,
    x="Quantity",
    y="Sales"
)

plt.title("Quantity vs Transaction Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

Most transactions are concentrated near the lower end of the chart.



Some transactions have extremely large positive or negative values.



Negative quantities and negative sales appear, which may represent returns or credit adjustments.



The chart does not show a simple, clear relationship because extreme values compress the majority of observations.

### Multivariate Analysis : It examines three or more variables together to understand patterns and relationship


```python
# Visualize calculated sales using quantity and price
sales_central = sales_data_clean.loc[
    sales_data_clean["Quantity"].between(1, 100)
    & sales_data_clean["Calculated_Sales"].between(0, 500)
]

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=sales_central,
    x="Quantity",
    y="Calculated_Sales",
    hue="Price",
    palette="viridis"
)

plt.title("Calculated Sales by Quantity and Price")
plt.xlabel("Quantity")
plt.ylabel("Calculated Sales")
plt.show()
```

```python
sales_data_clean["Calculated_Sales"] = (
    sales_data_clean["Quantity"] * sales_data_clean["Price"]
)
sales_data_clean["Calculated_Sales"].head()
```
