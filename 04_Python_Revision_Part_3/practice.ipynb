FUNCTIONS

```python
price  = 2500
discount_percentage = 15

def calculate_discount(price , discount_percentage):
     return price * (discount_percentage/100)

def calculate_final_price(price , discount_percentage):
     discount = calculate_discount(price,discount_percentage)
     return price - discount

print("Discount : ",calculate_discount(price , discount_percentage))
print("Final Price : ",calculate_final_price(price , discount_percentage))
```

LIST COMPREHENSION

```python
monthly_sales = [
    100000,
    150000,
    180000,
    220000,
    250000
]

sales_after_bonus = [round(sale * 1.10) for sale in monthly_sales]
print(sales_after_bonus)

high_performing_sales = [sale for sale in monthly_sales if sale >= 180000]
print(high_performing_sales)
```
