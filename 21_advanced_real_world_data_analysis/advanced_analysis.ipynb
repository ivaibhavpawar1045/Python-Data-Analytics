```python
# Date features
df["Order Year"] = df["order date (DateOrders)"].dt.year
df["Order Month"] = df["order date (DateOrders)"].dt.month
df["Order Month Name"] = df["order date (DateOrders)"].dt.month_name()
df["Order Day Name"] = df["order date (DateOrders)"].dt.day_name()
```

REGIONAL SEGEMNT PERFORMANCE

```python
region_analysis = df.groupby("Order Region").agg(
     Total_Sales = ("Sales" , "sum"),
     Total_Profit= ("Order Profit Per Order" , "sum"),
     Average_Sales= ("Sales" , "mean"),
     Average_Profit= ("Order Profit Per Order" , "mean"),
     Total_Orders = ("Order Id" , "nunique")
)
region_analysis.sort_values(
     by="Total_Orders",
     ascending=True
)
```

Western Europe has the highest total sales: ~5.89M

Western Europe has the highest average sales: ~217.43

Central America has 9,396 orders, making it another major region by order volume.

Northern Europe has the highest average profit among the displayed regions: ~23.84

CUSTOMER SEGMENT

```python
segment_analysis = df.groupby("Customer Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Order Profit Per Order", "sum"),
    Average_Sales=("Sales", "mean"),
    Average_Profit=("Order Profit Per Order", "mean"),
    Total_Orders=("Order Id", "nunique")
)
segment_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)
```

```python
segment_analysis["Total_Sales"].idxmax()
```

```python
segment_analysis["Total_Profit"].idxmax()
```

```python
segment_analysis["Total_Orders"].idxmax()
```

SALES CONTRIBUTION BY SEGMENT

```python
total_sales = segment_analysis["Total_Sales"].sum()
total_sales
```

```python
segment_analysis["Sales_Contribution_%"] = (
     segment_analysis["Total_Sales"]
     .div(total_sales)
     .mul(100)
)
segment_analysis["Sales_Contribution_%"]
```

PROFIT CONTRIBUTION BY SEGMENT

```python
segment_analysis["Profit_Contribution_%"] = (
    segment_analysis["Total_Profit"]
    .div(segment_analysis["Total_Profit"].sum())
    .mul(100)
)
```

```python
segment_analysis[
    ["Total_Sales", "Sales_Contribution_%", "Total_Profit", "Profit_Contribution_%"]
].sort_values(
    by="Profit_Contribution_%",
    ascending=False
)
```

Sales vs Profit Analysis

```python
category_analysis["Profit_Margin_%"] = (
    category_analysis["Total_Profit"]
    .div(category_analysis["Total_Sales"])
    .mul(100)
)
category_analysis.sort_values(
    by="Profit_Margin_%",
    ascending=False
)
```

Delivery Risk Analysis

```python
delivery_risk_region = df.groupby("Order Region")[
    "Late_delivery_risk"
].mean().mul(100)
delivery_risk_region.sort_values(
    ascending=False
)
```

Within-Region Segment Contribution

```python
region_segment = df.groupby(
    ["Order Region", "Customer Segment"]
)["Sales"].sum().reset_index()
```

```python
region_segment["Region_Total_Sales"] = (
    region_segment.groupby("Order Region")["Sales"]
    .transform("sum")
)
```

```python
region_segment["Sales_Contribution_%"] = (
    region_segment["Sales"]
    .div(region_segment["Region_Total_Sales"])
    .mul(100)
)
```

```python
dominant_segment = region_segment.loc[
    region_segment.groupby("Order Region")["Sales_Contribution_%"].idxmax()
]
```

```python
dominant_segment.sort_values(
    by="Sales_Contribution_%",
    ascending=False
)
```
