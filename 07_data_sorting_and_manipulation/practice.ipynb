SORT_VALUES ()

```python
import pandas as pd
```

```python
employee_df = pd.read_csv("employee.csv")
employee_df
```

```python
# SORT EMPLOYEES BY SALARY IN  ACSENDING ORDER
employee_df.sort_values(
     by=["Salary"]
     )
```

```python
# SORT EMPLOYEES BY SALARY IN  DESCENDING ORDER

employee_df.sort_values(
     by=["Salary"] ,
     ascending=False
     )
```

```python
# SORT EMPLOYEES BY EMPLOYEE NAME IN  ALPHABETICAL ORDER

employee_df.sort_values(
     by=["Employee"]
)
```

```python
employee_df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)
```

DROP_DUPLICATES ()

```python
import pandas as pd
```

```python
employee_data = {
     "Employee": ["Rahul", "Priya", "Rahul", "Amit", "Priya"],
    "Department": ["Sales", "HR", "Sales", "IT", "HR"],
    "Salary": [50000, 65000, 50000, 70000, 65000]
}

duplicate_df = pd.DataFrame(employee_data)

duplicate_df
```

```python
clean_df = duplicate_df.drop_duplicates()

clean_df
```

```python

employee_unique = duplicate_df.drop_duplicates(
     subset = ["Employee"]
)

employee_unique
```
