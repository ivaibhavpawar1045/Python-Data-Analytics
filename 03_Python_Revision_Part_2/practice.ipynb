DICTIONARIES

```python
customer = {
     "Customer ID" : 101,
     "Customer Name" : "Rohit Sharma",
     "City" : "Pune",
     "Total Orders" : 18,
     "Premium Customer" : True
}

print("Customer Details :",customer)

print("Customer Name :", customer["Customer Name"])

print("City Name :" , customer["City"])

customer["Total Orders"] = 22
print("Updated Total Orders :", customer["Total Orders"])

customer["Membership"] = "Gold"
print("Updated Details :",customer)

print("Keys :",customer.keys())

print("Values :", customer.values())

for key , value in customer.items():
     print(f"{key}:{value}")
```

SETS

```python
customer_purchases = [
    "Rahul",
    "Amit",
    "Rahul",
    "Sneha",
    "Priya",
    "Amit",
    "Vaibhav",
    "Sneha"
]

unique_customers = set(customer_purchases)

print("Unique Customers :", unique_customers)

print("Total Unique Customers :", len(unique_customers))

unique_customers.add("Rohit")

unique_customers.remove("Amit")

print("Updated Customers :", unique_customers)
```

INDEXING AND SLICING

```python
monthly_revenue = [
    210000,
    225000,
    238000,
    252000,
    261000,
    275000,
    288000,
    301000,
    315000,
    329000,
    340000,
    355000
]

print("First Quarter:", monthly_revenue[:3])

print("Second Quarter:", monthly_revenue[3:6])

print("Last Quarter:", monthly_revenue[-3:])

print("Alterante Month:", monthly_revenue[::2])

print("Reverse list:", monthly_revenue[::-1])

print("Second and Third Quarter:", monthly_revenue[3:9])
```

LOOPS (PROCESSING BUSINESS DATA EFFICIENTLY)

```python
monthly_sales = [
    125000,
    155000,
    98000,
    175000,
    145000,
    182000
]

sales_target = 150000

for sale in monthly_sales:
     print(sale)


for sale in monthly_sales:

    if sale >= sales_target:
        print(f"{sale} → Target Achieved")

    else:
        print(f"{sale} → Below Target")


successful_months = 0

for sale in monthly_sales:

    if sale >= sales_target:
        successful_months += 1

print("Successful Months :", successful_months)

          
```
