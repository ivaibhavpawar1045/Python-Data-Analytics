```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create business dataset
sales_data = pd.DataFrame({
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Clothing", "Clothing", "Clothing",
        "Furniture", "Furniture", "Furniture",
        "Food", "Food", "Food"
    ],
    "Region": [
        "North", "South", "West",
        "North", "South", "West",
        "North", "South", "West",
        "North", "South", "West"
    ],
    "Sales": [
        12000, 15000, 13500,
        9000, 11000, 10000,
        18000, 22000, 20000,
        7000, 8000, 7500
    ],
    "Profit": [
        1800, 2500, 2100,
        1000, 1500, 1200,
        3200, 4500, 3800,
        700, 900, 800
    ],
    "Customers": [
        20, 25, 22,
        18, 21, 19,
        30, 40, 35,
        15, 17, 16
    ]
})

sales_data.head()
```

### Advanced Seaborn: hue, palette and errorbar

```python
# Compare sales by category and region
plt.figure(figsize=(8,4))

sns.barplot(
     data=sales_data,
     x = "Category" ,
     y = "Sales" ,
     hue = "Region",
     palette="Set2",
    errorbar=None
)

plt.title("Sales by Category and Region")
plt.xlabel("Product Category")
plt.ylabel("Sales")
plt.legend(title = "Region")
plt.show()
```

```python
# Management wants to compare average profit by category and region.
plt.figure(figsize=(8,4))

sns.barplot(
     data=sales_data,
     x="Category",
     y="Profit",
     hue="Region",
     palette="Set2",
     errorbar=None
)

plt.title("Average Profit by Category and Region")
plt.xlabel("Product Category")
plt.ylabel("Profit")
plt.legend("Region")
plt.show()
```
