ADVANCED DATA ANALYSIS WITH PANDAS

```python
import pandas as pd
```

```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" ,
                 encoding="latin1")
df.head()
```

SALES CONTRIBUTION BY MARKET

```python
# find the sales by market
market_sales = (
     df.groupby("Market")["Sales"]
     .sum()
)
market_sales
```

```python
# find total sales
total_sales = df["Sales"].sum()
total_sales
```

```python
# sales contribution in market
sales_contribution = (
     market_sales
     .div(total_sales)
     .mul(100)
     .reset_index(name="Sales Contribution %")
)
sales_contribution
```

SALES CONTRIBUTION BY CUSTOMER SEGMENT

```python
# find customer segment sales
customer_segment_sales = (
     df.groupby("Customer Segment")["Sales"]
     .sum()
)
customer_segment_sales
```

```python
# find the total sales
total_sales = df["Sales"].sum()
total_sales
```

```python
sales_contribution = (
     customer_segment_sales
     .div(total_sales)
.mul(100).reset_index(name="Sales Contribution %")
)
sales_contribution
```

CONTRIBUTION WITHIN GROUP

```python
# find the market and customer segment sales
market_segment_sales = (
    df.groupby(
        ["Market", "Customer Segment"]
    )["Sales"]
    .sum()
    .reset_index()
)

market_segment_sales
```

```python
market_segment_sales["Market Total Sales"] = (
    market_segment_sales
    .groupby("Market")["Sales"]
    .transform("sum")
)

market_segment_sales["Market Total Sales"]
```

```python
market_segment_sales["Market Contribution %"] = (
    market_segment_sales["Sales"]
    .div(market_segment_sales["Market Total Sales"])
    .mul(100)
)

market_segment_sales
```

FIND THE DOMINANT SEGMENT IN EACH MARKET

```python
# find the highest contribution in each market
top_segment_by_market = market_segment_sales.loc[
     market_segment_sales.groupby("Market")["Market Contribution %"].idxmax()
]
top_segment_by_market
```

BUSINESS CHALLENGE

WHICH MARKET DEPENDS MOST HEAVILY ON ITS CONSUMER SEGMENT

```python
# filter consumer
consumer_market = market_segment_sales[
     market_segment_sales["Customer Segment"] == "Consumer"
]
consumer_market
```

```python
# find market with highest consumer contribution
consumer_market.loc[
    consumer_market["Market Contribution %"].idxmax()
]
```
