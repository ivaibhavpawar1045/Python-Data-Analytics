```python
import pandas as pd
import numpy as np
```

```python
# Create business dataset
sales =  np.array([
     1200,1500,1100,1800,
     900,1300,1250,1600,
     2000,2200,2100,2500
])
sales
```

```python
# reshape the array
regional_sales = sales.reshape(3,4)
regional_sales
```

```python
regional_sales.shape
```

```python
# reshape into 4 rows and 3 columns
sales.reshape(4,3)
```

```python
# Flatten array 
regional_sales.flatten()
```

```python
# management wants to calculate the total sales for each region
regional_sales.sum(axis=1)
```

### BOOLEAN OPERATIONS AND np.where()

```python
# Find sales that are at least 1,200 and below 2,000.
sales [
     (sales >= 1200) &
     (sales < 2000)
]
```

```python
# Find sales below 1,000 or at least 2,200.
sales [
     (sales < 1000) | 
     (sales >= 2200)
 ]
```

```python
# classify sales using np.where()
sales_status = np.where(
     sales >= 1800 ,
     "High" ,
     "Standard"
)
sales_status
```

```python
# count each category
np.unique(sales_status, return_counts=True)
```

```python
# Sales ≥ 2,000 → "Excellent"
# Sales ≥ 1,500 → "Good"
# Otherwise → "Needs Improvement"

sales_performance =np.select(
     [
          sales >= 2000 ,
          sales >= 1500
     ],
     [
          "Excellent",
          "Good"
     ],
     default="Need Improvemnet"
)
sales_performance
```

```python
np.unique(sales_performance, return_counts=True)
```

### BROADCASTING AND NUMPY WITH PANDAS

```python
# A 10% price increase across every sales value:
increased_sales = sales * 1.10
increased_sales
```

```python
regional_adjustments = np.array([1.10, 1.05, 1.15])

adjusted_regional_sales = regional_sales * regional_adjustments[:, np.newaxis]

adjusted_regional_sales
```

```python
# Convert NumPy results into a Pandas DataFrame
sales_df = pd.DataFrame(
     regional_sales ,
     index=["Region A" , "Region B" , "Region C"],
     columns=["Day 1"  , "Day 2" , "Day 3" , "Day 4"]
)
sales_df
```

```python
# add regional KPI
sales_df["Total_Sales"] = sales_df.sum(axis=1)

sales_df
```

```python
# Calculate contribution percentage using NumPy
sales_df["Contribution_%"] = (
     np.divide(
          sales_df["Total_Sales"],
          sales_df["Total_Sales"].sum()
     ) * 100
)
sales_df
```
