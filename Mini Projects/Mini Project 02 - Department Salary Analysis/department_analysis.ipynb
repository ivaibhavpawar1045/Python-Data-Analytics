SECOND MINI PROJECT : DEPARTMENT SALES ANALYSIS

```python
import pandas as pd
```

```python
employee_data = {
    "Employee": [
        "Rahul","Priya","Amit","Sneha","Rohan",
        "Neha","Karan","Pooja","Vikram","Anjali"
    ],
    "Department": [
        "Sales","HR","IT","Finance","IT",
        "Sales","HR","Finance","IT","Sales"
    ],
    "Salary": [
        50000,65000,70000,55000,80000,
        62000,68000,72000,90000,58000
    ],
    "Experience":[
        2,5,4,3,6,
        4,5,7,8,3
    ]
}

employee_df = pd.DataFrame(employee_data)

employee_df
```

```python
employee_df.head()
```

```python
employee_df.info()
```

```python
# department wise total salary
department_total_salary = employee_df.groupby("Department")["Salary"].sum()

department_total_salary
```

```python
# department wise average salary
department_average_salary = employee_df.groupby("Department")["Salary"].mean()

department_average_salary
```

```python
# department wise employee count
department_count_salary = employee_df.groupby("Department")["Salary"].count()

department_count_salary
```

```python
# department wise maximum salary
department_max_salary = employee_df.groupby("Department")["Salary"].max()

department_max_salary
```

```python
# department wise minimum salary
department_min_salary = employee_df.groupby("Department")["Salary"].min()

department_min_salary
```

```python
# professional calculation
employee_df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "count", "min", "max"]
)
```

```python
# employee with highest salary
employee_df.sort_values(
     by = "Salary",
     ascending=False
).head(1)
```

```python
# employee with lowest salary
employee_df.sort_values(
     by="Salary"
).head(1)
```
