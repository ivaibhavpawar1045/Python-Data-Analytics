MAP() , APPLY() , REPLACE() AND LAMBADA() FUNCTIONS

```python
import pandas as pd
```

```python
employee_df = pd.DataFrame({
     "Employee" :["Rahul" , "Priya" , "Amit" , "Sneha"],
     "Department" :["Sales" , "HR" , "IT" , "Finance"],
     "Salary" :[50000,65000,70000,55000]
})

employee_df
```

MAP()

```python
# converting Department Names into Short Codes
employee_df["Department_Code"] = employee_df["Department"].map({
     "Sales" : "S",
     "HR" : "H",
     "IT" : "I",
     "Finance" : "F"
})

employee_df
```

```python
# convert salary into grades
employee_df["Salary_Grades"] = employee_df["Salary"].map({
     50000 : "C",
     55000 : "C",
     65000 : "B",
     70000 : "A"
})

employee_df
```

REPLACE()

```python
# replacing department name
employee_df["Department"].replace(
     "HR" ,
     "Human Resources"
)
```

```python
# replacing multiple column names
employee_df["Department"].replace({
     "Sales" : "Sales Team",
     "IT" : "Information Technology"
})
```

```python
#replacing salary
employee_df["Salary"].replace(
     50000,52000
)
```

```python
# replacing entire DataFrame
employee_df.replace({
     "Sales" : "Sales Team",
     70000 : 75000})
```

APPLY()

```python
# increase every salary by 10 %
employee_df["Salary"].apply(
     lambda salary: salary * 1.10
)
```

```python
# convert employee names to uppercase
employee_df["Employee"].apply(
     lambda name:name.upper()
)
```

```python
# salary status
employee_df["Salary"].apply(
     lambda salary : "High"
     if salary >= 60000 else "Low"
)
```

LAMBDA()

```python
# increase salary by 20 %
employee_df["Salary"].apply(
     lambda salary : salary*1.20
)
```

```python
# employee name length
employee_df["Employee"].apply(
     lambda name : len(name)
)
```

```python

employee_df  = pd.DataFrame({
     "Employee" : ["Rahul " , "Priya" , "Amit" , "Sneha"],
     "Salary" : [50000,65000,70000,55000],
     "Experience" : [2,5,4,1]
})
employee_df
```

```python
# Experience Bonus
employee_df["Bonus"] = employee_df["Experience"].apply(
     lambda bonus : bonus * 5000
)
employee_df
```
