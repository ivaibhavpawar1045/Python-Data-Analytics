PIVOT TABLE AND CROSSTAB

```python
import pandas as pd
```

```python
employee_data = {
    "Employee": [
        "Rahul","Priya","Amit","Sneha",
        "Rohan","Neha","Karan","Pooja"
    ],
    "Department": [
        "Sales","HR","IT","Finance",
        "IT","Sales","HR","Finance"
    ],
    "City": [
        "Pune","Mumbai","Pune","Delhi",
        "Mumbai","Pune","Delhi","Mumbai"
    ],
    "Salary": [
        50000,65000,70000,55000,
        80000,62000,68000,72000
    ]
}

employee_df = pd.DataFrame(employee_data)

employee_df
```

PIVOT_TABLE()

```python
# average salary by department
employee_df.pivot_table(
     values="Salary",
     index="Department",
     aggfunc="mean"
)
```

```python
# total salary by department
employee_df.pivot_table(
     values="Salary",
     index="Department",
     aggfunc="sum"
)
```

MULTIPLE AGGREGATIONS

```python
employee_df.pivot_table(
     values="Salary",
     index="Department",
     aggfunc=["mean" , "sum" , "min" , "max" , "count"]
)
```

MULTIPLE INDEX

```python
employee_df.pivot_table(
     values="Salary",
     index=["Department" , "City"],
     aggfunc="mean"
)
```

PIVOT()

```python
sales_df = pd.DataFrame({
    "Month": ["Jan", "Jan", "Feb", "Feb"],
    "Product": ["Laptop", "Mouse", "Laptop", "Mouse"],
    "Sales": [50000, 8000, 60000, 9000]
})

sales_df
```

```python
sales_df.pivot(
     columns="Product",
     index="Month",
     values="Sales"
)
```

CROSSTAB()

```python
pd.crosstab(
     employee_df["Department"],
     employee_df["City"]
)
```
