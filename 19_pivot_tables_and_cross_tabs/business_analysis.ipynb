PIVOT TABLES AND CROSS TAB BUSINESS ANALLYSIS

```python
import pandas as pd
```

```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" , encoding="latin1")
df.head()
```

PIVOT TABLE

```python
# how much sales does each customer segment generate in each market.
market_segment_pivot = pd.pivot_table(
     df ,
     values="Sales" ,
     index= "Market",
     columns="Customer Segment" ,
     aggfunc="sum")
market_segment_pivot
```

```python
# for every market and customer segment ,  show me both profit and sales
market_segment_pivot = pd.pivot_table(
     df ,
     values=["Sales" , "Order Profit Per Order"] ,
     index= "Market",
     columns="Customer Segment" ,
     aggfunc="sum")
market_segment_pivot
```

```python
# waht is total sales for each market , regardless of customer segment
sales_pivot = pd.pivot_table(
     df ,
     values="Sales" ,
     index="Market" ,
     columns="Customer Segment" ,
     aggfunc="sum" ,
     margins=True
)
sales_pivot
```

CROSS TABULATION

```python
# Cross Tabulation
market_segment_count = pd.crosstab(
     df["Market"] , 
     df["Customer Segment"]
)
market_segment_count
```

```python
     # add grand total to crosstab
market_segment_count = pd.crosstab(
     df["Market"] ,
     df["Customer Segment"],
     margins=True
)
market_segment_count
```
