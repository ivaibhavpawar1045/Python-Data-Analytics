```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create small business dataset
sales_data = pd.DataFrame({
     "Category" : [
          "Electronics" , "Electronics" ,
          "Clothing" , "Clothing" ,
          "Furniture" , "Furniture",
          "Food" , "Food" 
     ] ,
     "Sales" : [
          12000,15000,
          9000 ,11000,
          18000 , 22000,
          7000 , 8000
     ]
})
sales_data
```

```python
plt.figure(figsize=(8,4))

sns.barplot(data=sales_data, x="Category" , y="Sales")

plt.title("Average Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Sales (₹)")
plt.show()
```

```python
sales_data.groupby("Category")["Sales"].mean()
```

### Count plots and histograms

```python
plt.figure(figsize=(8, 4))

sns.countplot(
    data=sales_data,
    x="Category"
)

plt.title("Number of Records by Category")
plt.xlabel("Product Category")
plt.ylabel("Record Count")
plt.show()
```

```python
# HISTOGRAM
plt.figure(figsize=(8, 4))

sns.histplot(
    data=sales_data,
    x="Sales",
    bins=5
)

plt.title("Distribution of Sales")
plt.xlabel("Sales (₹)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Count the number of records per category and find which category appears most frequently.
category_counts = sales_data["Category"].value_counts()
category_counts
```

```python
# Then identify the category with the highest count:
print("Most Frequent Category : " , category_counts.idxmax())
print("Record Count :" , category_counts.max())
```

### Scatter plots and box plots

```python
# SCATTER PLOT     
sales_data["Profit"] = [
    1800, 2500,
    1000, 1500,
    3200, 4500,
    700, 900
]


plt.figure(figsize=(8, 4))

sns.scatterplot(
    data=sales_data,
    x="Sales",
    y="Profit",
    hue="Category"         # it assigns different colors to the observations for categories , helping comparing groups
)

plt.title("Sales vs Profit by Category")
plt.xlabel("Sales (₹)")
plt.ylabel("Profit (₹)")
plt.show()
```

```python
# BOX PLOT
plt.figure(figsize=(8, 4))

sns.boxplot(
    data=sales_data,
    x="Category",
    y="Sales"
)

plt.title("Sales Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Sales (₹)")
plt.show()
```

```python
# Calculate the profit margin for each record and compare average margins by category.
sales_data["Profit_Margin_%"] = (
    sales_data["Profit"] / sales_data["Sales"] * 100
)

sales_data.groupby("Category")["Profit_Margin_%"].mean().sort_values(
    ascending=False
)
```
