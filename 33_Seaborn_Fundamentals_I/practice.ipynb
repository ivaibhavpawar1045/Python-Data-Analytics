```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create small business dataset
sales_data = pd.DataFrame({
     "Category" : [
          "Electronics" , "Electronics" ,
          "Clothing" , "Clothing" ,
          "Furniture" , "Furniture",
          "Food" , "Food" 
     ] ,
     "Sales" : [
          12000,15000,
          9000 ,11000,
          18000 , 22000,
          7000 , 8000
     ]
})
sales_data
```

```python
plt.figure(figsize=(8,4))

sns.barplot(data=sales_data, x="Category" , y="Sales")

plt.title("Average Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Sales (₹)")
plt.show()
```

```python
sales_data.groupby("Category")["Sales"].mean()
```
