STRING OPERATIONS IN PANDAS

```python
import pandas as pd
```

```python
employee_df = pd.DataFrame({
     "Employee" : [" Rahul " , "PRIYA" , "amit" , "Sneha"],
     "Email" : ["Rahul@Company.COM","PRIYA@Company.COM","amit@Company.COM","Sneha@Company.COM"]
})

employee_df
```

```python
# Convert name to lowercase
employee_df["Employee"].str.lower()
```

```python
# Convert name to uppercase
employee_df["Employee"].str.upper()
```

```python
# Convert name in title
employee_df["Employee"].str.title()
```

```python
# Removes Extra spaces from name
employee_df["Employee"].str.strip()
```

CONTAINS()

```python
# find name which contains a
employee_df["Employee"].str.contains("a")
```

```python
# find emails contains company
employee_df["Email"].str.contains("Company")
```

```python
# Ignore the case sensitivity
employee_df["Email"].str.contains(
     "company",
     case=False
)
```

STARTWITH() , ENDSWITH() , REPLACE() , SPLIT()

```python
# Checks the string startwith....
employee_df["Employee"].str.startswith("R")
# it shows all false because the " Rahul " contains extra spaces
```

```python
# Checks string endswith...
employee_df["Email"].str.endswith(".COM")
```

```python
# Replace the part of string
employee_df["Email"].str.replace(
     "Company",
     "Google"
)
```

```python
# Replace the employee name
employee_df["Employee"].str.replace(
     "Rahul",
     "Rohit"
)
```

```python
players_df = pd.DataFrame({
     "Full Name" : [
          "Rohit Sharma",
          "Harmanpreet Kaur",
          "Virat Kohli",
          "Smriti Mandhana"
     ]
})
```

```python
# use to split the strin
players_df["Full Name"].str.split(" ")
```

```python
# create the first name column and access the first name from full name
players_df["First Name"] = players_df["Full Name"].str.split().str[0]

players_df
```

```python
players_df["Last Name"] = players_df["Full Name"] .str.split().str[1]

players_df
```
