HANDLING MISSING VALUES

```python
import pandas as pd
import numpy as np
```

```python
employee_data = {
      "Employee": ["Rahul", "Priya", "Amit", "Sneha"],
    "Department": ["Sales", "HR", np.nan, "Finance"],
    "Salary": [50000, np.nan, 70000, 55000]
}

employee_df = pd.DataFrame(employee_data)
```

```python
employee_df
```

ISNULL ()

```python
employee_df.isnull()
```

```python
employee_df.isnull().sum()
```

NOTNULL()

```python
employee_df.notnull()
```

```python
employee_df.notnull().sum()
```

FILLNA()

```python
employee_df.fillna(0)
```

```python
employee_df["Salary"].fillna(0)
```

```python
employee_df["Department"].fillna("Unknown")
```

```python
employee_df.fillna(
     {
          "Department" : "Unknown",
          "Salary" : 0
     }
)
```

DROPNA()

```python
employee_df.dropna()
```

```python
employee_df.dropna(axis=1)
```

REAL BUSINESS PROBLEM DATASET

```python
import pandas as pd
import numpy as np
```

```python
employee_data = {
    "Employee": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Neha"],
    "Department": ["Sales", "HR", np.nan, "Finance", "IT", np.nan],
    "Salary": [50000, np.nan, 70000, 55000, np.nan, 62000],
    "Experience": [2, 5, 4, np.nan, 6, 3]
}

hr_df = pd.DataFrame(employee_data)

hr_df
```

```python
hr_df.isnull().sum()
```

```python
hr_df["Department"].fillna("Unknown")
```

```python
hr_df["Salary"].fillna(0)
```

```python
hr_df["Experience"].fillna(0)
```

```python
clear_hr_df  = hr_df.fillna({
     "Department" : "Unknown",
     "Salary" : 0,
     "experience" : 0
})

clear_hr_df
```

```python
clear_hr_df.isnull().sum()
```
