DATA TRANFORMATION AND FEATURE ENGINEERING

```python
# Extract date types
df["Order Year"] = df["order date (DateOrders)"].dt.year
df["Order Month"] = df["order date (DateOrders)"].dt.month
df["Order Month Name"] = df["order date (DateOrders)"].dt.month_name()
df["Order Day Name"] = df["order date (DateOrders)"].dt.day_name()
df[
     [
          "order date (DateOrders)",
          "Order Year",
          "Order Month",
          "Order Month Name",
          "Order Day Name",
     ]
].head()
```

```python
# Create customer full name
df["Customer Name"] = df["Customer Fname"] + " " +  df["Customer Lname"]
df.head()
```

```python
# create profit status 
df["Profit Status"] = df["Order Profit Per Order"].apply(
     lambda x : "Profit" if x >= 0 else "Loss"
)
df.head()
# df["Profit Status"].value_counts()
```

```python
# check delivery risk
df["Delivery Status"] = df["Late_delivery_risk"].map({
     0 : "No Risk",
     1 : "Late delivery risk"
}
)
df[["Late_delivery_risk" , "Delivery Status"]].head()
```

```python
df.head()
```

BUSINESS ANALYSIS

```python
# total sales and profit
df["Total_Sales"] = df["Sales"].sum()
df["Total_Profit"] = df["Order Profit Per Order"].sum()
# df[["Total_Sales" , "Total_Profit"]]
df["Total_Sales"].sum() , df["Total_Profit"].sum()
```

```python
# Sales and profit by Category
category_analysis = (
    df.groupby("Category Name")[["Sales", "Order Profit Per Order"]]
    .sum()
    .sort_values("Sales", ascending=False)
)
category_analysis
```

```python
# highest category and most prifitable
category_analysis["Sales"].idxmax() , category_analysis["Order Profit Per Order"].idxmax()
```

```python
# Regional Performance
region_analysis = (
    df.groupby("Order Region")[["Sales", "Order Profit Per Order"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

region_analysis.head(10)
```

```python
# highest sales region and profit region
region_analysis["Sales"].idxmax() ,region_analysis["Order Profit Per Order"].idxmax()
```

```python
# customer segment analysis
segment_analysis = (
    df.groupby("Customer Segment")[["Sales", "Order Profit Per Order"]]
    .sum()
    .sort_values("Sales", ascending=False)
)
segment_analysis
```

```python
# Top 10 customers
top_customers = (
     df.groupby("Customer Name")["Sales"]
     .sum().sort_values(ascending=False)
     .head(10)
)
top_customers
```

```python
profit_status = df["Profit Status"].value_counts()

profit_status
```

```python
# total loss
loss_df = df[df["Profit Status"] == "Loss"]
loss_df["Order Profit Per Order"].sum()
```

```python
loss_by_category = (
    loss_df.groupby("Category Name")["Order Profit Per Order"]
    .sum()
    .sort_values(ascending = False)
)

loss_by_category
```

```python
# Monthly sales trends
monthly_sales = (
    df.groupby(["Order Year", "Order Month"])["Sales"]
    .sum()
)
# monthly_sales
monthly_sales.idxmax()
```
