MINI PROJECT 03 - HR EMPLOYEE DATA TRANSFORMATIONS

```python
import pandas as pd
```

```python
employee_df = pd.read_csv("employee_data.csv")
employee_df
```

```python
# Department Code
employee_df["Department_Code"]= employee_df["Department"].map({
     "Sales" : "S",
     "HR" : "H",
     "IT" : "I",
     "Finance" : "F"
})
employee_df
```

```python
# Replace Department Name
employee_df["Department_name"] = employee_df["Department"].replace({
     "HR" : "Human Resources" , 
     "IT" : "Information Technology"
})
employee_df
```

```python
# Increase salary by 10 %
employee_df["Updated_Salary"] = employee_df["Salary"].apply(
     lambda salary : salary * 1.10
)
employee_df
```

```python
# Salary Status
employee_df["Salary_Status"] = employee_df["Salary"].apply(
     lambda salary : "High"
     if salary >= 60000 else "Low"
)
employee_df
```

```python
# Increase Bonus
employee_df["Bonus"] = employee_df["Salary"].apply(
     lambda bonus : bonus * 5000
)
employee_df
```
