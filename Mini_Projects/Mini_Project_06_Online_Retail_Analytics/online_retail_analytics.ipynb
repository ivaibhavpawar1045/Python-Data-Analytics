HANDLE MISSING VALUES

```python
clean_df["Description"] = clean_df["Description"].fillna("Unknown Product")
clean_df["Description"].isnull().sum()
```

Create Transaction Type

```python
clean_df["Transaction Type"] = np.where(
     clean_df["Invoice"].astype(str).str.startswith("C") , 
     "Cancellation" ,
     "Sale"
)
clean_df["Transaction Type"].value_counts()
```

```python
# create revenue
clean_df["Revenue"] = clean_df["Quantity"] * clean_df["Price"]
clean_df[["Quantity" , "Price" , "Revenue"]]
```

```python
# Create Date feature
clean_df["Year"] = clean_df["InvoiceDate"].dt.year
```

```python
clean_df["Month"] = clean_df["InvoiceDate"].dt.month
```

```python
clean_df["Month Name"] = clean_df["InvoiceDate"].dt.month_name()
```

```python
clean_df["Day Name"] = clean_df["InvoiceDate"].dt.day_name()
```

COUNTRY ANALYSIS

```python
# country analysis 
country_revenue = clean_df.groupby("Country")["Revenue"].sum()
country_revenue.sort_values(ascending=False).head(10)
```

```python
country_analysis = clean_df.groupby("Country").agg(
     total_revenue = ("Revenue" , "sum"),
     total_quantity = ("Quantity" , "sum"),
     total_transactions = ("Invoice" , "nunique")
)
country_analysis.sort_values(
     by = "total_revenue",
     ascending=False
)
```

```python
# Revenue Contribution %
total_country_revenue = country_analysis["total_revenue"].sum()
```

```python
country_analysis["Revenue Contribution %"] = (
     country_analysis["total_revenue"].div(total_country_revenue)
     .mul(100)
)
```

```python
country_analysis.sort_values(
     by = "Revenue Contribution %" ,
     ascending = False
).head(10)
```

```python
# Customer Analysis
customer_df = clean_df.dropna(subset=["Customer ID"]).copy()
```

```python
customer_analysis = customer_df.groupby("Customer ID").agg(
     total_revenue = ("Revenue" , "sum") ,
     total_quantity = ("Quantity" , "sum") ,
     total_transactions = ("Invoice" , "nunique") 
)
```

```python
customer_analysis.sort_values(
     by = "total_revenue" ,
     ascending = False
).head(10)
```

```python
# Customer Revenue Contribution %
total_customer_revenue = customer_analysis["total_revenue"].sum()
```

```python
customer_analysis["Customer Revenue Contribution %"] = (
     customer_analysis["total_revenue"]
     .div(total_customer_revenue)
     .mul(100)
)
```

```python
customer_analysis.sort_values(
     by = "Customer Revenue Contribution %" ,
     ascending = False
).head(10)
```

Product Analysis

```python
product_analysis = clean_df.groupby("StockCode").agg(
     total_revenue = ("Revenue" , "sum"),
     total_quantity = ("Quantity" , "sum"),
     total_transactions = ("Invoice" , "nunique")
)
```

```python
product_analysis.sort_values(
     by = "total_revenue" ,
     ascending = False
).head(10)
```

```python
# Add Product Description
product_lookup = clean_df[
    ["StockCode", "Description"]
].drop_duplicates(subset="StockCode")
```

```python
product_analysis = product_analysis.reset_index().merge(
    product_lookup,
    on="StockCode",
    how="left"
)
```

```python
product_analysis.sort_values(
    by="total_revenue",
    ascending=False
).head(10)
```

```python
# product revenue contribution %
total_product_revenue = product_analysis["total_revenue"].sum()
```

```python
product_analysis["Product Revenue Contribution %"] = (
     product_analysis["total_revenue"]
     .div(total_product_revenue)
     .mul(100)
)
```

```python
product_analysis.sort_values(
     by = "total_revenue" ,
     ascending = False
).head(10)
```
