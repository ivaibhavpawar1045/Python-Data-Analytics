```python
#Remove duplicates
clean_df = clean_df.drop_duplicates()
clean_df.shape
```

HANDLE MISSING VALUES

```python
clean_df["Description"] = clean_df["Description"].fillna("Unknown Product")
clean_df["Description"].isnull().sum()
```

Create Transaction Type

```python
clean_df["Transaction Type"] = np.where(
     clean_df["Invoice"].astype(str).str.startswith("C") , 
     "Cancellation" ,
     "Sale"
)
clean_df["Transaction Type"].value_counts()
```

```python
# create revenue
clean_df["Revenue"] = clean_df["Quantity"] * clean_df["Price"]
clean_df[["Quantity" , "Price" , "Revenue"]]
```

```python
# Create Date feature
clean_df["Year"] = clean_df["InvoiceDate"].dt.year
```

```python
clean_df["Month"] = clean_df["InvoiceDate"].dt.month
```

```python
clean_df["Month Name"] = clean_df["InvoiceDate"].dt.month_name()
```

```python
clean_df["Day Name"] = clean_df["InvoiceDate"].dt.day_name()
```
