REMOVE EXTRASPACES

```python
# remove extra spaces
customer_df["Full_Name"] = customer_df["Full_Name"].str.strip()
customer_df
```

```python
customer_df["City"] = customer_df["City"].str.strip()
customer_df
```

CONVERT INTO TITLES

```python
customer_df["Full_Name"] = customer_df["Full_Name"].str.title()
customer_df
```

```python
customer_df["City"] = customer_df["City"].str.title()
customer_df
```

FORMAT EMAILS

```python
customer_df["Email"] = customer_df["Email"].str.lower()
customer_df
```

CHANGE DOMAIN

```python
customer_df["Email"] = customer_df["Email"].str.replace(
     "company" , "techcorp"
)
customer_df
```

EXTRACT NAME

```python
customer_df["First_Name"] = customer_df["Full_Name"].str.split().str[0]
customer_df
```

```python
customer_df["Last_Name"] = customer_df["Full_Name"].str.split().str[1]
customer_df
```
