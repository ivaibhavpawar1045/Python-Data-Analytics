# DAY 38 : MULTIVARIATE EXPLORATORY DATA ANALYSIS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("online_retail_II_cleaned.xlsx")
df.shape
```

```python
# ==========================================
# DAY 38: MULTIVARIATE EDA
# ==========================================

# 1. SELECT IMPORTANT NUMERICAL VARIABLES

analysis_df = df[
    ["Quantity", "Price", "Sales"]
].copy()

print("Selected Dataset:")
display(analysis_df.head())


# 2. CORRELATION MATRIX

correlation_matrix = analysis_df.corr()

print("Correlation Matrix:")
display(correlation_matrix)
```

```python
# 3. CORRELATION HEATMAP

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Multivariate Correlation Heatmap")
plt.tight_layout()
plt.show()


# 4. MULTIVARIATE SCATTER PLOT

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df.sample(
        n=min(2000, len(df)),
        random_state=42
    ),
    x="Quantity",
    y="Sales",
    hue="Is_Cancelled",
    alpha=0.6
)

plt.title("Quantity vs Sales by Cancellation Status")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()

```

```python

# 5. PAIR PLOT

sample_df = df.sample(
    n=min(1000, len(df)),
    random_state=42
)

sns.pairplot(
    sample_df[
        ["Quantity", "Price", "Sales", "Is_Cancelled"]
    ],
    hue="Is_Cancelled",
    diag_kind="hist"
)

plt.suptitle(
    "Pairwise Relationships in Online Retail Data",
    y=1.02
)

plt.show()


# 6. FINAL CORRELATION SUMMARY

print("Correlation with Sales:")

display(
    correlation_matrix["Sales"]
    .sort_values(ascending=False)
)
```
