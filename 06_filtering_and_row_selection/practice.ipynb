MULTIPLE CONDITION (AND & and OR |)

```python
employee_df[
     (employee_df["Salary"]> 50000) &
     (employee_df["Department"] == "IT")
]
```

```python
employee_df[
     (employee_df["Salary"]<60000) &
     (employee_df["Department"]== "Finance") 
]
```

```python
employee_df[
     (employee_df["Department"] == "Sales") |
     (employee_df["Department"]== "HR") 
]
```

```python
employee_df[
     (employee_df["Salary"]>60000) |
     (employee_df["Department"]== "Finance") 
]
```

ISIN()

```python
employee_df[
     employee_df["Department"].isin(["IT" , "HR"])
]
```

```python
employee_df[
     employee_df["Department"].isin(["Sales" , "Finance"])
]
```

```python
employee_df[
     employee_df["Department"].isin(["IT" , "Sales" , "HR"])
]
```

BETWEEN ()

```python
import pandas as pd
```

```python
employee_df = pd.read_csv("employee.csv")
employee_df
```

```python
employee_df[
     employee_df["Salary"].between(50000,65000)]
```

```python
employee_df[
     employee_df["Salary"].between(55000,70000)
     ]
```

```python
employee_df[
     employee_df["Salary"].between(60000,70000)
]
```

LOC ()

```python
employee_df.loc[
     employee_df["Salary"] > 60000 ,
     ["Employee" , "Salary"]
]
```

```python
employee_df.loc[
     employee_df["Department"] == "HR",
     ["Employee" , "Department"]
]
```

```python
employee_df.loc[
     employee_df["Department"].isin(["Sales","Finance"]) ,
     ["Employee" , "Salary"]
]
```

```python
employee_df.loc[
     : ,
     ["Employee" , "Salary"]
]
```
