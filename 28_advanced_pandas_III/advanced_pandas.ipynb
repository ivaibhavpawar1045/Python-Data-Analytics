### transform()


```python
# calculate the total sales generated by its market.
df["Market_Total_Sales"] = df.groupby(
    "Market"
)["Sales"].transform("sum")
```

```python
df[[
    "Market",
    "Sales",
    "Market_Total_Sales"
]].head(10)
```

```python
# percentage contribution to its market:
df["Market_Sales_Contribution_%"] = (
    df["Sales"]
    .div(df["Market_Total_Sales"])
    .mul(100)
)
```

```python
df[[
    "Market",
    "Sales",
    "Market_Total_Sales",
    "Market_Sales_Contribution_%"
]].head(10)
```

```python
df.groupby("Market")["Sales"].sum()
```

### Group Level Ranking

```python
# top products within each market
product_market_sales = df.groupby(
     ["Market" , "Product Name"]
).agg(
     Total_Sales = ("Sales" , "sum")
).reset_index()
product_market_sales.head()
```

```python
product_market_sales["Market_Product_Rank"] = (
    product_market_sales
    .groupby("Market")["Total_Sales"]
    .rank(
        method="dense",
        ascending=False
    )
)
```

```python
product_market_sales.head(10)
```

```python
top_products = product_market_sales[
     product_market_sales["Market_Product_Rank"] <= 3
]
top_products.sort_values(
     ["Market" , "Market_Product_Rank"]
).head(10)
```

### shift()

```python
monthly_sales = df.groupby(
    [
        df["order date (DateOrders)"].str[:10],
        "Market"
    ]
)["Sales"].sum().reset_index()

monthly_sales.columns = [
    "Order_Date",
    "Market",
    "Total_Sales"
]
```

```python
monthly_sales["Order_Date"] = pd.to_datetime(
    monthly_sales["Order_Date"] , format="mixed"
)
```

```python
monthly_sales = monthly_sales.sort_values(
    ["Market", "Order_Date"]
)
monthly_sales.head(15)
```

```python
monthly_sales["Previous_Sales"] = (
    monthly_sales
    .groupby("Market")["Total_Sales"]
    .shift(1)
)
monthly_sales.head(15)
```

```python
daily_sales = monthly_sales.copy()
```

```python
daily_sales.head()
```

```python
daily_sales = daily_sales.rename(columns={
    "Total_Sales": "Daily_Sales",
    "Previous_Sales": "Previous_Day_Sales"
})
```

```python
daily_sales.head(10)
```

### rolling() + Cumulative Calculations



## rolling()

```python
daily_sales = daily_sales.sort_values(
    ["Market", "Order_Date"]
)
```

```python
# Management wants a 3-day moving average of sales for each market.
daily_sales["3_Day_Moving_Avg"] = (
    daily_sales
    .groupby("Market")["Daily_Sales"]
    .transform(
        lambda x: x.rolling(3).mean()
    )
)
```

```python
daily_sales[
    ["Order_Date", "Market", "Daily_Sales", "3_Day_Moving_Avg"]
].head(10)
```

##   Cumulative Sales

```python
# Calculate cumulative sales within each market.
daily_sales["Cumulative_Sales"] = (
    daily_sales
    .groupby("Market")["Daily_Sales"]
    .cumsum()
)
```

```python
daily_sales[
    [
        "Order_Date",
        "Market",
        "Daily_Sales",
        "Cumulative_Sales"
    ]
].head(10)
```

## 3️⃣ Cumulative Maximum

```python
daily_sales["Cumulative_Max_Sales"] = (
    daily_sales
    .groupby("Market")["Daily_Sales"]
    .cummax()
)
```

```python
daily_sales[
    [
        "Order_Date",
        "Market",
        "Daily_Sales",
        "Cumulative_Sales",
        "Cumulative_Max_Sales"
    ]
].head(10)
```

```python
daily_sales["Sales_Growth"] = (
    daily_sales["Daily_Sales"]
    - daily_sales["Previous_Day_Sales"]
)
```

```python
daily_sales["Growth_Status"] = np.where(
    daily_sales["Sales_Growth"] > 0,
    "Increased",
    "Decreased"
)
```

```python
daily_sales["Growth_Status"].value_counts()
```

```python
daily_sales.loc[
    daily_sales["Sales_Growth"].idxmax(),
    [
        "Order_Date",
        "Market",
        "Daily_Sales",
        "Previous_Day_Sales",
        "Sales_Growth"
    ]
]
```
