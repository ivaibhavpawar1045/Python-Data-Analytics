### MultiIndex + Advanced groupby() + transform()

```python
market_segment = df.groupby(
     ["Market" , "Customer Segment"]
)["Sales"].sum()
market_segment
```

```python
market_segment_df = market_segment.to_frame(
     name="Total_Sales"
)
market_segment_df
```

### transform()


```python
# calculate the total sales generated by its market.
df["Market_Total_Sales"] = df.groupby(
    "Market"
)["Sales"].transform("sum")
```

```python
df[[
    "Market",
    "Sales",
    "Market_Total_Sales"
]].head(10)
```

```python
# percentage contribution to its market:
df["Market_Sales_Contribution_%"] = (
    df["Sales"]
    .div(df["Market_Total_Sales"])
    .mul(100)
)
```

```python
df[[
    "Market",
    "Sales",
    "Market_Total_Sales",
    "Market_Sales_Contribution_%"
]].head(10)
```
