EXPLORATORY DATA ANALYSIS

```python
import pandas as pd
```

```python
clean_df = pd.read_csv("superstore_clean.csv")
clean_df.head()
```

```python
clean_df["Order Date"] = pd.to_datetime(
     clean_df["Order Date"],
     format="mixed"
)
clean_df
```

```python
clean_df.info()
```

```python
# Calculate total sales and profit
total_sales = clean_df["Sales"].sum
total_profit = clean_df["Profit"].sum
total_sales , total_profit
```

```python
# Calculate average sales and profit
average_sales = clean_df["Sales"].sum()
average_profit = clean_df["Profit"].sum()
average_sales , average_profit
```

```python
# Compare sales and profit together 
category_performance = clean_df.groupby(
     ["Category"]
)[["Sales" , "Profit"]].sum()
category_performance
```

```python
# find most profitable category
category_performance["Profit"].idxmax() , category_performance["Profit"].max()
```

```python
# regional performance
regional_performance = clean_df.groupby(
     ["Region"]
)[["Sales" , "Profit"]].sum()
regional_performance
```

```python
# find most profitable region
regional_performance["Profit"].idxmax() , regional_performance["Sales"].max()
```

```python
# loss making transactions
loss_orders = clean_df[
     clean_df["Profit"] < 0
]
loss_orders
```

```python
loss_orders.shape[0]
```

```python
# check how many loss
loss_orders["Profit"].sum()   # this show negative result for absolute we use absolute 'abs'
abs(loss_orders["Profit"].sum())
```

```python
# which category has the most loss
loss_by_category = loss_orders.groupby(
    "Category"
)["Profit"].sum()

loss_by_category
```

```python
loss_by_category.idxmin() ,  loss_by_category.min()
```

TOP 10 CUSTOMERS

```python
top_customers = clean_df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_customers
```
