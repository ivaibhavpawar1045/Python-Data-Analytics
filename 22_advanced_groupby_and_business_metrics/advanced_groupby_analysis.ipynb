DAY 22 ADVANCED GROUPBY AND BUSINESS METRICS

```python
import pandas as pd
```

```python
df = pd.read_csv("Data_Co_Supply_Chain_Dataset.csv" ,
encoding="latin")
df.shape
```

```python
df.columns
```

ADVANCED REGION ANALYSIS

```python
region_analysis = df.groupby("Order Region").agg(
     Total_sales = ("Sales" , "sum"),
     Total_profit = ("Order Profit Per Order" , "sum"),
     Average_sales = ("Sales" , "mean"),
     Average_profit = ("Order Profit Per Order" , "mean"),
     Total_orders = ("Order Id" , "nunique")
)
region_analysis.head()
```

PROFIT MARGIN %

```python
region_analysis["Profit_Margin_%"] = (
     region_analysis["Total_profit"]
     .div(region_analysis["Total_sales"])
     .mul(100)
)
region_analysis.sort_values(
     by="Profit_Margin_%" ,
     ascending=False
)
```

SALES PER ORDER

```python
region_analysis["Sales_Per_Order"] = (
     region_analysis["Total_sales"]
     .div(region_analysis["Total_orders"])
)
region_analysis.sort_values(
     by="Sales_Per_Order",
     ascending=False
)
```

```python
region_analysis["Sales_Per_Order"].idxmax()
```

```python
region_analysis["Sales_Per_Order"].idxmin()
```

## REGION + CUSTOMER SEGMENT

```python
region_segment = df.groupby(
     ["Order Region" , "Customer Segment"]
)["Sales"].sum().reset_index()
region_segment.head(10)
```

```python
region_segment.sort_values(
     by="Sales" ,
     ascending=False
).head(10)
```

WITHIN SEGMENT CONTRIBUTION

```python
region_segment["Region_Total_Sales"] = (
     region_segment.groupby("Order Region")["Sales"]
     .transform("sum")
)
region_segment["Sales_Contribution_%"] = (
     region_segment["Sales"]
     .div(region_segment["Region_Total_Sales"])
     .mul(100)
)
region_segment.sort_values(
     by="Sales_Contribution_%" ,
     ascending=False
).head(10)
```

DOMINANT SEGMENT IN EVERY REGION

```python
dominat_segment = region_segment.loc[
     region_segment.groupby("Order Region")[
          "Sales_Contribution_%"
     ].idxmax()
]
dominat_segment.sort_values(
     by="Sales_Contribution_%" ,
     ascending = False
)
```

## CUSTOMER VALUE ANALYSIS

```python
#Customer-level aggregation
customer_analysis = df.groupby("Customer Id").agg(
     Total_sales = ("Sales" , "sum"),
     Total_orders = ("Order Id" , "nunique"),
     Average_order_value = ("Sales" , "mean")
     
)
customer_analysis.sort_values(
     by="Total_sales" ,
     ascending=False
).head(10)
```

Customer Sales Contribution

```python
total_customer_sales = customer_analysis["Total_sales"].sum()

customer_analysis["Sales_Contribution_%"] = (
     customer_analysis["Total_sales"]
     .div(total_customer_sales).mul(100)
)
customer_analysis.sort_values(
     by="Sales_Contribution_%" ,
     ascending=False
).head(10)
```

Customer Value: Order Value Comparison

```python
customer_analysis["Average_order_value"].idxmax()
```

```python
customer_analysis.loc[
    customer_analysis["Average_order_value"].idxmax()
]
```

High-Value Customer Identification


```python
high_value_customers = customer_analysis[
     (customer_analysis["Total_sales"] > 5000) &
     (customer_analysis["Total_orders"] >= 10) 
]
high_value_customers.sort_values(
     by="Total_sales" ,
     ascending = False
).head(10)
```

## PRODUCT PERFORMANCE

```python
product_analysis = df.groupby("Product Name").agg(
     Total_sales = ("Sales" , "sum"),
     Total_quantity = ("Order Item Quantity" , "sum"),
     Total_orders = ("Order Id" , "nunique")
)
product_analysis.sort_values(
     by="Total_sales" ,
     ascending=False
).head(10)
```

```python
# sales contribution
total_product_sales = product_analysis["Total_sales"].sum()
product_analysis["Sales_Contribution_%"] = (
     product_analysis["Total_sales"]
     .div(total_product_sales)
     .mul(100)
)
product_analysis.sort_values(
     by="Sales_Contribution_%" ,
     ascending=False
).head(10)
```

```python
product_analysis["Sales_Contribution_%"].idxmax()
```
