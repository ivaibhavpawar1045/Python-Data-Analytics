# DAY 35 : ADVANCED SEABORN VISUALIZATION AND MULTIVARIATE ANALYSIS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_data = pd.DataFrame({
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Clothing", "Clothing", "Clothing",
        "Furniture", "Furniture", "Furniture",
        "Food", "Food", "Food"
    ],
    "Region": [
        "North", "South", "West",
        "North", "South", "West",
        "North", "South", "West",
        "North", "South", "West"
    ],
    "Sales": [
        12000, 15000, 13500,
        9000, 11000, 10000,
        18000, 22000, 20000,
        7000, 8000, 7500
    ],
    "Profit": [
        1800, 2500, 2100,
        1000, 1500, 1200,
        3200, 4500, 3800,
        700, 900, 800
    ],
    "Customers": [
        20, 25, 22,
        18, 21, 19,
        30, 40, 35,
        15, 17, 16
    ],
    "Discount": [
        5, 10, 8,
        10, 15, 12,
        8, 12, 10,
        2, 5, 3
    ]
})

sales_data.head()
```

```python
sales_data.shape
```

```python
sales_data.info()
```

```python
sns.pairplot(
     data=sales_data ,
     vars=["Sales" , "Profit" , "Customers" , "Discount"],
     hue="Category"
)
plt.show()
```

```python
sns.pairplot(
     data=sales_data,
     vars=["Sales" , "Profit" , "Discount"],
     hue="Region"
)
plt.show()
```

### jointplot() and relplot()

```python
sns.jointplot(
    data=sales_data,
    x="Sales",
    y="Profit",
    kind="scatter",
    hue="Category"
)

plt.show()
```

```python
sns.relplot(
    data=sales_data,
    x="Sales",
    y="Profit",
    hue="Region",
    col="Category",
    kind="scatter"
)

plt.show()
```

```python
# Create a relational plot comparing Customers vs Sales, colored by Region and split into panels by Category
sns.relplot(
    data=sales_data,
    x="Customers",
    y="Sales",
    hue="Region",
    col="Category",
    kind="scatter"
)

plt.show()
```

```python
sales_data["Profit_Margin_%"] = (
    sales_data["Profit"]
    / sales_data["Sales"]
    * 100
)
```

```python
# calculates average profit margin by category and ranks the categories from highest to lowest.

# Calculate average profit margin by category
category_margin = (
    sales_data.groupby("Category")["Profit_Margin_%"]
    .mean()
    .sort_values(ascending=False)
)

category_margin
```
