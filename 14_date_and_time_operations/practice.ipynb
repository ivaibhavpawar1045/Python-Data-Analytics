DATE AND TIME OPERATIONS IN PANDAS

```python
import pandas as pd
```

```python
sales_df = pd.DataFrame({
    "Order_ID":[101,102,103,104],
    "Customer":["Rahul","Priya","Amit","Sneha"],
    "Order_Date":[
        "2026-08-01",
        "2026-08-03",
        "2026-08-05",
        "2026-08-06"
    ],
    "Sales":[25000,18000,32000,21000]
})

sales_df
```

CORRECT DATA TYPE

```python
sales_df.info()
```

```python
sales_df["Order_Date"] = pd.to_datetime(
     sales_df["Order_Date"]
)
sales_df.info()
```

EXTRACT YEAR , MONTH , DAY

```python
sales_df["Year"] = sales_df["Order_Date"].dt.year
sales_df
```

```python
sales_df["Month"] = sales_df["Order_Date"].dt.month
sales_df
```

```python
sales_df["Day"] = sales_df["Order_Date"].dt.day
sales_df
```

DAY NAME , MONTH NAME

```python
sales_df["Day_Name"] = sales_df["Order_Date"].dt.day_name()
sales_df
```

```python
sales_df["Month_Name"] = sales_df["Order_Date"].dt.month_name()
sales_df
```

DATE FILTERING

```python
# Orders after August 3
sales_df[ sales_df["Order_Date"] > "2026-08-03"]
```

```python
# Orders in August
sales_df[
     sales_df["Month"] == 8
]
```

```python
# Orders from 2026
sales_df[
     sales_df["Year"] == 2026
]
```

```python
# Weekend Orders
sales_df[sales_df["Day_Name"].isin(
    [ "Saturday",
     "Sunday"]
)]
```

```python
# Total sales in August
sales_df[
     sales_df["Month"] == 8
]["Sales"].sum()
```

```python
# Average sales
sales_df["Sales"].mean()
```

```python
# highest sales
sales_df["Sales"].max()
```

```python
# lowest sales
sales_df["Sales"].min()
```

```python
sales_df.sort_values(
     by = "Sales" ,
     ascending=False
)
```
